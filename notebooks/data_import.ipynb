# Download data

```python
dir = Path("../data/deer-iyhon")
dir.mkdir(parents=True, exist_ok=True)

rf = Roboflow(api_key=ROBOFLOW_API_KEY)

project = rf.workspace("aidan-dyer").project("deer-iyhon")
version = project.version(6)

dataset = version.download(
    "yolo26"
)
```

```python
dataset = sv.DetectionDataset.from_yolo(
    images_directory_path="../data/deer-iyhon/train/images",
    annotations_directory_path="../data/deer-iyhon/train/labels",
    data_yaml_path="../data/deer-iyhon/data.yaml"
)
print(dataset.classes)
```

```python

def visualize_sample(dataset: sv.DetectionDataset, index: int = 0):
    """Visualize an image and its bounding boxes from a supervision DetectionDataset."""
    image_path, image, annotations = dataset[index]

    box_annotator = sv.BoxAnnotator()
    label_annotator = sv.LabelAnnotator()

    labels = [
        dataset.classes[class_id]
        for class_id in annotations.class_id
    ]

    annotated_image = box_annotator.annotate(scene=image.copy(), detections=annotations)
    annotated_image = label_annotator.annotate(
        scene=annotated_image, detections=annotations, labels=labels
    )

    plt.figure(figsize=(8, 8))
    plt.imshow(annotated_image[:, :, ::-1])  # BGR -> RGB
    plt.axis("off")
    plt.title(image_path.split("/")[-1])
    plt.show()

# Example usage
visualize_sample(dataset, index=205)
```

```python
def images_by_bbox_count(dataset: sv.DetectionDataset, top_n: int = 5):
    """Return the top_n images with the most bounding boxes, sorted descending."""
    counts = [
        (image_path, len(annotations.class_id))
        for image_path, _, annotations in dataset
    ]
    counts.sort(key=lambda x: x[1], reverse=True)
    return counts[:top_n]

# Example usage
top_images = images_by_bbox_count(dataset, top_n=5)
for path, count in top_images:
    print(f"{count} boxes — {path}")
```

```python
def visualize_top_bbox_images(dataset: sv.DetectionDataset, top_n: int = 3):
    """Visualize the top_n images with the most bounding boxes."""
    top_images = images_by_bbox_count(dataset, top_n=top_n)
    # Build a lookup from image_path to index in dataset
    path_to_index = {path: i for i, (path, _, _) in enumerate(dataset)}

    for path, count in top_images:
        idx = path_to_index[path]
        print(f"{count} boxes — {path}")
        visualize_sample(dataset, index=idx)
```

```python
visualize_top_bbox_images(dataset)
```

```python
# visualize_sample(dataset, index=205)
def crop_deers(index, dataset=dataset):
    image = cv2.imread(dataset[index][0])
    cropped_images = []

    for x1, y1, x2, y2 in dataset[index][2].xyxy:
        crop = image[int(y1):int(y2), int(x1):int(x2)].copy()
        cropped_images.append(crop)

    return cropped_images
```

```python
def visualize_image(image):
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    plt.imshow(image_rgb)
    plt.axis("off")
    plt.show()
```

```python
visualize_sample(dataset, 1000)
images = crop_deers(1000)
for image in images:
    visualize_image(image)
    
```
